### Exploratory Data Analysis and Model Prototyping

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load MovieLens 100k Sample
df = pd.read_csv("../data/u.data", sep='\t', names=['user_id', 'movie_id', 'rating', 'timestamp'])
print(f"Dataset Shape: {df.shape}")
```

## 📊 Data Visualization
Understanding the interaction density and popularity bias.

```python
plt.figure(figsize=(10, 5))
sns.countplot(x='rating', data=df, palette='Reds_r')
plt.title("User Rating Distribution")
plt.show()
```

## 📈 Model Convergence
Visualizing the Binary Cross Entropy Loss across 10 training epochs.

```python
epochs = list(range(10))
loss = [0.6806, 0.0087, 0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 0.0000, 0.0000]

plt.figure(figsize=(10, 5))
plt.plot(epochs, loss, marker='o', color='#E50914')
plt.title("Neural Tower Convergence (BCE Loss)")
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.show()
```
